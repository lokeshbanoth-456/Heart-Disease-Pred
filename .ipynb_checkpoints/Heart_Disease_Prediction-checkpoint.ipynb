# Data Preprocessing and Cleaning 

```python
df_encode=pd.get_dummies(df,drop_first=True)
```

```python
df_encode
```

```python
df_encode=df_encode.astype(int)
```

```python
df_encode
```

# Feature Engineering and Extaction

```python
df_encode['Age'] = df['Age']
```

```python
df_encode['Age'].head()
```

```python
df_encode['Age_Group'] = pd.cut(
    df_encode['Age'],
    bins=[0, 30, 45, 60, float('inf')],
    labels=['Young', 'Adult', 'Middle', 'Senior'],
    include_lowest=True
)
```

```python
df_encode
```

```python
from sklearn.preprocessing import StandardScaler

numerical_columns = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']

scaler = StandardScaler()

df_encode[numerical_columns] = scaler.fit_transform(df_encode[numerical_columns])
```

```python
df_encode.head()
```

```python
df_encode['High_BP'] = (df['RestingBP'] >= 140).astype(int)
df_encode['High_Cholesterol'] = (df['Cholesterol']>=240).astype(int)
df_encode['Senior'] = (df['Age'] >= 60).astype(int)
df_encode['Age_BP'] = df['Age'] * df['RestingBP']
```

```python
df_encode['BP_Cholesterol'] = (
    df['RestingBP'] *
    df['Cholesterol']
)

df_encode['Exercise_Risk'] = (
    df_encode['ExerciseAngina_Y'] *
    df['Oldpeak']
)
```
